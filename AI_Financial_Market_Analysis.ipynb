# **ANALYSIS**

## 1) How much amt of companies spent on R & D?

```python
df.head()
```

```python
df.groupby('Company')['R&D_Spending_USD_Mn'].sum()    #values in million
```

```python
df.groupby('Company')['R&D_Spending_USD_Mn'].sum()/1000    #values in billion
```

```python
print("Company's spending for R & D in $Bn")

RD = df.groupby('Company')['R&D_Spending_USD_Mn'].sum()/1000

RD
```

```python
# TO DRAW THE BAR GRAPH FOR THE AMT ON COMPANIES SPENT ON R & D
plt.bar(RD.index, RD.values, color=['cyan','black','magenta'])
plt.title('Company-wise R&D Spending ($Bn)')
plt.xlabel('Company')
plt.ylabel('R&D Spending ($Bn)')
plt.xticks(rotation=45)
plt.show()
```

# 2) Revenue earned by companies

```python
df.head()
```

```python
df.groupby('Company')['AI_Revenue_USD_Mn'].sum()/1000
```

```python
print("Company's AI Revenue_USD_Bn :")

Rev = df.groupby('Company')['AI_Revenue_USD_Mn'].sum()/1000

Rev
```

```python
# Draw the bar plot to show the revenue of the companies
plt.bar(Rev.index, Rev.values, color=['green','yellow','red'])
plt.title('Company-wise AI Revenue ($Bn)')
plt.xlabel('Company')
plt.ylabel('AI Revenue ($Bn)')
plt.xticks(rotation=45)
plt.show()
```

```python
# to show both amt and revenue of companies in single bar plots by using subplots
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.bar(RD.index, RD.values, color=['steelblue','teal','navy'])
plt.title('Company-wise R&D Spending ($Bn)')
plt.xlabel('Company')
plt.ylabel('R&D Spending ($Bn)')
plt.xticks(rotation=45)

plt.subplot(1,2,2)
plt.bar(Rev.index, Rev.values, color=['orange','gold','crimson'])
plt.title('Company-wise AI Revenue ($Bn)')
plt.xlabel('Company')
plt.ylabel('AI Revenue ($Bn)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Expenditure vs Revenue year by year

```python
df.head()
```

```python
spend=df.groupby('Year')['R&D_Spending_USD_Mn'].sum()
spend
```

```python
#showing amt on R&D
plt.plot(spend.index, spend.values, color='black')
plt.title("Combined R&D Spending Year-by-Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_Mn")
plt.show()
```

```python
revenue=df.groupby('Year')['AI_Revenue_USD_Mn'].sum()
revenue
```

```python
#show revenue earned
plt.plot(revenue.index, revenue.values, color='red')
plt.title("Combined AI Revenue Year-by-Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_Mn")
plt.show()
```

```python
plt.plot(revenue.index, revenue.values, color='red')
plt.plot(spend.index, spend.values, color='black')
plt.title("Combined R&D Spending and AI Revenue Year-by-Year",fontsize=10)
plt.xlabel("Year")
plt.ylabel("Amount in USD_Mn")
plt.legend(['AI Revenue','R&D Spending'])
plt.show()
```
